## Del 1: Fra grenseverdi til differanse

Den deriverte beskriver endring. I kjemi skriver vi den ofte som $\dfrac{dc}{dt}$ (reaksjonsfart) eller $\dfrac{d\mathrm{pH}}{dV}$ (titrerkurve).

<div class="viktig">

Den deriverte er definert som en grenseverdi:

$$f'(x)=\lim_{h\rightarrow 0}\frac{f(x+h)-f(x)}{h}$$

En datamaskin kan ikke bruke en uendelig liten $h$. Vi bruker derfor en liten, endelig steglengde. Det kalles **framoverdifferansen**:

$$f'(x)\approx\frac{f(x+h)-f(x)}{h}$$

## Del 5: Derivasjon forsterker støy

Vi lager kunstige konsentrasjonsdata fra $c(t) = e^{-0{,}1t}$ og legger til litt målestøy med standardavvik 0,005 mol/L.

<div class="diskuter">

**Tipp først:** Støyen er bare omtrent 1 % av konsentrasjonen. Hvor mye tror du den påvirker reaksjonsfarten $-dc/dt$? Kjør koden og se.

</div>

```python
rng = np.random.default_rng(1)
t = np.linspace(0, 30, 61)
c = np.exp(-0.1 * t)
c_noisy = c + rng.normal(0, 0.005, len(t))

fig, axes = plt.subplots(1, 2, figsize=(13, 4))
axes[0].plot(t, c_noisy, "o", markersize=4)
axes[0].set_ylabel("c (mol/L)")
axes[1].plot(t, -np.gradient(c_noisy, t), "o-", markersize=4, label="From noisy data")
axes[1].plot(t, 0.1 * c, label="Exact")
axes[1].set_ylabel("-dc/dt (mol/(L s))")
axes[1].legend()
for ax in axes:
    ax.set_xlabel("t (s)")
plt.show()
```

<div class="viktig">

Derivasjon deler små forskjeller mellom nabopunkter på et lite intervall. Da blir tilfeldig målestøy **forsterket**. Vurder alltid rådata og den deriverte sammen. Glatting kan hjelpe, men den endrer også dataene.

</div>
