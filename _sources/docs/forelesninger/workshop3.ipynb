    ```{code-block} Python
    import numpy as np
    import matplotlib.pyplot as plt

    # Konstanter og variabler
    m = 1    # Masse i kg
    k = 2    # Fjærkonstant
    x_eq = 0 # Likevektsposisjon i m
    v0 = 0   # Startfart i m/s
    x0 = 1   # Startposisjon i m

    # Tidssteg
    dt = 0.001
    tid = 40        # Tid i s
    N = int(tid/dt) # Intervaller

    # Initialisering
    v = np.zeros(N+1)
    x = np.zeros(N+1)
    t = np.zeros(N+1)

    v[0] = v0
    x[0] = x0

    # Integrasjonsløkke
    for i in range(N):
        a = -k/m*(x[i]-x_eq)
        # Eulers metode
        v[i+1] = v[i] + a*dt
        x[i+1] = x[i] + v[i]*dt
        t[i+1] = t[i] + dt

    plt.plot(t,x)
    plt.show()
    ```
