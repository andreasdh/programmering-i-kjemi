### F6. Valg av t-test i SciPy

En forsker sammenlikner reaksjonshastigheten for **to uavhengige grupper** med separate reaksjonsblandinger: én gruppe med katalysator (`med`) og én uten (`uten`). Hypotesen som skal testes, er at katalysatoren **øker** den gjennomsnittlige reaksjonshastigheten. Vi ønsker ikke å forutsette at de to gruppene har lik varians.

Hvilken linje gjennomfører en egnet **ensidig** t-test?

A. `test = stats.ttest_rel(med, uten, alternative="greater")`  
B. `test = stats.ttest_ind(med, uten, equal_var=False, alternative="less")`  
C. `test = stats.ttest_1samp(med, popmean=0, alternative="greater")`  
D. `test = stats.ttest_ind(med, uten, equal_var=False, alternative="greater")`

*Anta at `from scipy import stats` er kjørt.*

```{dropdown} Fasit med begrunnelser – flervalg
- **F1: A.** `&` krever at begge temperaturbetingelsene er oppfylt, og `dropna` fjerner rader der ledningsevnen mangler. B bruker `|` (eller), C fjerner mangler fra feil kolonne, D utelukker 30 °C.
- **F2: C.** `groupby` samler alle målinger ved 25 °C. `mean()` ser bort fra den manglende verdien: $(590+600)/2=595$.
- **F3: A.** Bare parene ved 20 og 25 °C tilfredsstiller begge betingelsene. Verdien ved 30 °C er for høy.
- **F4: C.** $\mathrm{RSD}=100\cdot s/\bar{x}$. B mangler multiplikasjon med 100.
- **F5: B.** For $\kappa=aT+b$ er $\Delta\kappa=a\Delta T=a\cdot8$. Konstantleddet faller bort når vi beregner *endringen*. Enheten til `slope` er (µS/cm)/°C.
- **F6: D.** Gruppene er uavhengige, så vi velger en uparet test. `equal_var=False` gir Welchs t-test, og `greater` undersøker om gjennomsnittet i `med` er høyere enn i `uten`.
```
