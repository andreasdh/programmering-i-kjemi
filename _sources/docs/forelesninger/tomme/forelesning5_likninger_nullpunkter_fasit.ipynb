## Repetisjon: Koffein i energidrikker (15 til 20 minutter)

Ifølge etiketten inneholder en energidrikk 32 mg koffein per 100 mL. To laboratorier, A og B, har analysert koffeininnholdet i **de samme seks boksene** med HPLC. Her er et utsnitt av datasettet:

| can | lab | caffeine_mg_100mL |
|---|---|---|
| 1 | A | 31.2 |
| 1 | B | 31.8 |
| 2 | A | 32.5 |
| 2 | B | 33.0 |
| ... | ... | ... |

Kjør cella nedenfor for å lage hele datarammen.

```python
data = pd.DataFrame({
    "can": [1, 2, 3, 4, 5, 6] * 2,
    "lab": ["A"] * 6 + ["B"] * 6,
    "caffeine_mg_100mL": [31.2, 32.5, 30.8, 31.9, 31.4, 32.0,
                          31.8, 33.0, 31.5, 32.5, 32.0, 32.7],
})
data
```

<div class="oppgave">

**Del a) Beskriv dataene.** Beregn antall målinger, gjennomsnitt, standardavvik og RSD for hvert laboratorium. Er presisjonen omtrent lik i de to laboratoriene?

</div>

```python
summary = data.groupby("lab")["caffeine_mg_100mL"].agg(["count", "mean", "std"])
summary["rsd_percent"] = summary["std"] / summary["mean"] * 100
summary
```

<div class="oppgave">

**Del b) Stemmer etiketten?** Bruk målingene fra laboratorium A. Beregn et 95 % konfidensintervall og utfør en passende t-test mot 32 mg/100 mL. Hva kan du konkludere, og hva kan du *ikke* konkludere?

</div>

```python
lab_A = data[data["lab"] == "A"]["caffeine_mg_100mL"]

ci = stats.t.interval(0.95, len(lab_A) - 1, loc=lab_A.mean(), scale=stats.sem(lab_A))
test = stats.ttest_1samp(lab_A, popmean=32)

print(ci)
print(f"p = {test.pvalue:.2f}")
```

<div class="oppgave">

**Del c) Er laboratoriene uenige?** Velg en t-test for å sammenlikne A og B, og begrunn valget. Prøv gjerne også den andre testen. Hvorfor blir p-verdiene så forskjellige? Er forskjellen mellom laboratoriene kjemisk viktig?

Tips: `data[data["lab"] == "B"].sort_values("can")` gir målingene i samme rekkefølge som boksene.

</div>

```python
lab_A = data[data["lab"] == "A"].sort_values("can")["caffeine_mg_100mL"].values
lab_B = data[data["lab"] == "B"].sort_values("can")["caffeine_mg_100mL"].values

paired = stats.ttest_rel(lab_B, lab_A)
independent = stats.ttest_ind(lab_B, lab_A, equal_var=False)

print(f"Mean difference B - A: {np.mean(lab_B - lab_A):.2f} mg/100 mL")
print(f"Paired: p = {paired.pvalue:.6f}")
print(f"Independent: p = {independent.pvalue:.2f}")
```

<div class="viktig">

- **b)** 32 mg/100 mL ligger inne i konfidensintervallet, og $p = 0{,}20$. Dataene gir ikke grunnlag for å si at etiketten er feil. Det betyr **ikke** at vi har vist at den er riktig.
- **c)** Begge laboratoriene har målt **de samme boksene**, så dataene er parede. Koffeininnholdet varierer fra boks til boks, men B gir hele tiden omtrent 0,6 mg/100 mL mer enn A. Den parede testen ser denne systematiske forskjellen ($p < 0{,}0001$), mens den uavhengige testen drukner den i variasjonen mellom boksene ($p = 0{,}10$).

</div>

## Del 1: Likninger som nullpunktsproblemer

<div class="viktig">

Enhver likning $g(x) = h(x)$ kan skrives om til et **nullpunktsproblem**:

$$f(x) = g(x) - h(x) = 0$$

Mange likninger i kjemi har ingen analytisk løsning. Da finner vi nullpunktet numerisk, steg for steg.

</div>

### Eksempel: pH i 0,010 M eddiksyre
Fra massebalanse og ladningsbalanse (på tavla) får vi en funksjon av $h = [\mathrm{H_3O^+}]$:

$$f(h) = h - \frac{C K_a}{h + K_a} - \frac{K_w}{h}$$

pH finner vi der $f(h) = 0$.

```python
C = 0.010
Ka = 1.75e-5
Kw = 1.0e-14

def charge_balance(h):
    return h - C * Ka / (h + Ka) - Kw / h

h = np.logspace(-7, -2, 500)
plt.semilogx(h, charge_balance(h))
plt.axhline(0, color="gray")
plt.xlabel("[H$_3$O$^+$] (mol/L)")
plt.ylabel("f(h)")
plt.show()
```

</div>

## Del 2: Repetisjon av halveringsmetoden

<div class="oppgave">

**Les og forklar:** Diskuter koden nedenfor i par. Én forklarer linje 1 til 8, den andre resten. Svar på disse spørsmålene sammen:

1. Hvorfor sjekker vi `f(a)*f(m) < 0`?
2. Hvorfor bruker vi en toleranse i stedet for `f(m) != 0`?
3. Hva skjer hvis `f(a)` og `f(b)` har samme fortegn?
4. Hvorfor har funksjonen et maksimalt antall iterasjoner?

</div>

```python
def bisection(f, a, b, tol=1e-10, max_iter=100):
    i = 0
    m = (a + b) / 2

    while i < max_iter and abs(f(m)) > tol:
        if f(a) * f(m) < 0:
            b = m
        elif f(b) * f(m) < 0:
            a = m
        m = (a + b) / 2
        i = i + 1

    if i == max_iter:
        print("Maks antall iterasjoner er nådd.")
    return m, i

h_root, iterations = bisection(charge_balance, 1e-7, 1e-2)
print(f"pH = {-np.log10(h_root):.3f} etter {iterations} iterasjoner")
```

<div class="viktig">

Halveringsmetoden er **robust**: Har vi et fortegnsskifte, finner den alltid et nullpunkt. Men den er **treg**, fordi intervallet bare halveres for hver iterasjon.

</div>

## Del 3: Newtons metode

<div class="bilde-hoyre" style="width:50%">

![Tangenter i Newtons metode](bilder/newton_tangent.png)

</div>

**Idé:** Vi starter i en gjetning $x_0$, tegner tangenten og ser hvor den krysser x-aksen. Der er neste gjetning.

Vi utleder metoden på tavla.

<div class="viktig">

**Newtons metode**

$$x_{n+1} = x_n - \frac{f(x_n)}{f'(x_n)}$$

Vi gjentar til $|f(x_n)|$ er mindre enn en toleranse. Metoden trenger en **startverdi** og den **deriverte** $f'(x)$.

</div>

### Newtons metode i Python
Vi tester på $f(x) = x^2 - x - 2$, som har nullpunktet $x = 2$.

```python
def f(x):
    return x**2 - x - 2

def fder(x):
    return 2*x - 1

x = 5
while abs(f(x)) > 1e-8:
    x = x - f(x) / fder(x)

print(x)
```

<div class="oppgave">

**Prøv selv:** Hvor fort går det? Legg til en teller `i`, og skriv ut `i`, `x` og `f(x)` for hver iterasjon. Hvor mange iterasjoner trengs? Hva skjer med antall riktige desimaler for hver iterasjon?

</div>

```python
x = 5
i = 0
while abs(f(x)) > 1e-8:
    x = x - f(x) / fder(x)
    i = i + 1
    print(i, x, f(x))
```

<div class="viktig">

Nær nullpunktet blir antall riktige desimaler omtrent **doblet** for hver iterasjon. Det kalles **kvadratisk konvergens**. Halveringsmetoden vinner bare omtrent én desimal for hver tredje iterasjon.

</div>

## Del 4: pH med Newtons metode

Nå bruker vi Newtons metode på eddiksyreproblemet.

<div class="oppgave">

**Prøv selv:**

1. Deriver $f(h) = h - \dfrac{C K_a}{h + K_a} - \dfrac{K_w}{h}$ for hånd, og skriv funksjonen `charge_balance_der(h)`.
2. Bruk tilnærmingen $h_0 = \sqrt{K_a C}$ som startverdi. Hvor mange iterasjoner trenger Newtons metode? Sammenlikn med halveringsmetoden.

</div>

```python
def charge_balance_der(h):
    return 1 + C * Ka / (h + Ka)**2 + Kw / h**2

h = np.sqrt(Ka * C)
i = 0
while abs(charge_balance(h)) > 1e-12:
    h = h - charge_balance(h) / charge_balance_der(h)
    i = i + 1

print(f"pH = {-np.log10(h):.3f} etter {i} iterasjoner")
```

**Diskuter:** Hvorfor er $\sqrt{K_a C}$ en god startverdi? Hvorfor er det en god idé å bruke kjemikunnskap til å velge startverdi?

</div>

## Del 5: Når feiler Newtons metode?

Vi prøver $f(x) = x^3 - 2x + 2$ med startverdi $x_0 = 0$.

<div class="diskuter">

**Tipp først:** Hva skjer? Programmet stopper etter seks iterasjoner, slik at det ikke går i evig løkke.

</div>

```python
def p(x):
    return x**3 - 2*x + 2

def pder(x):
    return 3*x**2 - 2

x = 0
for i in range(6):
    x = x - p(x) / pder(x)
    print(i + 1, x)
```

<div class="viktig">

Newtons metode kan feile når

- startverdien ligger for langt unna nullpunktet, slik at iterasjonene går i ring eller havner ved et annet nullpunkt
- $f'(x_n) = 0$, slik at tangenten er vannrett og vi deler på null
- funksjonen ikke har noe nullpunkt i nærheten

</div>

<div class="oppgave">

**Finn feilen:** Funksjonen nedenfor kan gå i evig løkke. Rett den slik at den har et maksimalt antall iterasjoner og gir en advarsel, slik som `bisection`. Test den på `p` med startverdi 0 og startverdi −2.

</div>

```python
def newton(f, fder, x, tol=1e-10, max_iter=50):
    i = 0
    while abs(f(x)) > tol and i < max_iter:
        x = x - f(x) / fder(x)
        i = i + 1
    if i == max_iter:
        print("Maks antall iterasjoner er nådd. Prøv en annen startverdi.")
    return x

print(newton(p, pder, 0))
print(newton(p, pder, -2))
```

</div>

## Del 6: Ferdige løsere (hvis tid)

`root_scalar` fra SciPy har både halveringsmetoden og Newtons metode. Resultatet forteller om metoden konvergerte, og hvor mange iterasjoner som trengtes.

```python
from scipy.optimize import root_scalar

result = root_scalar(charge_balance, x0=np.sqrt(Ka * C), fprime=charge_balance_der, method="newton")
print(result.converged, result.iterations, -np.log10(result.root))
```
