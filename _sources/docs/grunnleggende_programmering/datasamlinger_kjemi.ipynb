Kjemiske data kommer sjelden som ett enkelt tall. Et spektrum inneholder absorbans ved mange bølgelengder. En titrering gir pH-verdier for en serie tilsatte volum. En simulering kan gi posisjonen og energien til mange partikler ved mange tidspunkt. For å arbeide med slike data må vi samle verdiene i en hensiktsmessig struktur.

Vi har flere måter å organisere data på i Python. Her er en kort oversikt over de viktigste datasamlingene:

1. **Lister** er fleksible samlinger av like eller ulike data.
2. **Arrayer** er samlinger av tall som vi kan utføre elementvise beregninger på.
3. **Tupler** er ordnede samlinger som ikke kan endres etter at de er opprettet.
4. **Dictionarier** lagrer verdier med meningsfulle nøkler i stedet for bare tallindekser.

Vi har allerede sett hvordan lister fungerer. Nå skal vi se nærmere på de tre andre datasamlingene og hvordan de kan brukes i kjemi.

## Arrayer

En *array* er en datasamling som er spesielt godt egnet for tall. Arrayer er sentrale når vi skal analysere måleserier, gjøre den samme beregningen på mange verdier eller representere data i rader og kolonner.

For å bruke arrayer importerer vi biblioteket NumPy. Det er vanlig å gi biblioteket kortnavnet `np`:

```{code-block} Python
import numpy as np
```

### Forskjellen mellom lister og arrayer

Vi begynner med et eksempel som illustrerer en viktig forskjell. Tenk deg at to kjemiske forbindelser bidrar til absorbansen ved tre bølgelengder:

```{code-block} Python
import numpy as np

absorbans_A_liste = [0.10, 0.18, 0.25]
absorbans_B_liste = [0.05, 0.07, 0.09]

print("Lister:", absorbans_A_liste + absorbans_B_liste)

absorbans_A = np.array(absorbans_A_liste)
absorbans_B = np.array(absorbans_B_liste)

print("Arrayer:", absorbans_A + absorbans_B)
```

Når to lister adderes, legges den andre lista til slutten av den første. Vi får én lengre liste. Når to arrayer med samme form adderes, legges verdiene sammen element for element. Den første absorbansverdien i `absorbans_A` legges til den første i `absorbans_B`, og så videre.

I dette eksempelet kan arrayaddisjonen tolkes kjemisk: Dersom Beer–Lamberts lov gjelder og forbindelsene ikke påvirker hverandre, kan de to forbindelsenes absorbansbidrag summeres ved hver bølgelengde. Koden er enkel, men den bygger altså på en modell av hvordan signalene kombineres.

### Behandle arraydata

I motsetning til lister bruker vi vanligvis ikke metoder som `append` og `remove` når vi arbeider med arrayer. Vi kan derimot hente ut elementer ved hjelp av indekser, akkurat som med lister. Indekseringen starter på 0.

```{code-block} Python
import numpy as np

bolgelengde = np.array([400, 450, 500, 550, 600, 650, 700])
absorbans = np.array([0.05, 0.08, 0.22, 0.71, 0.43, 0.16, 0.07])

print(bolgelengde[0])    # første bølgelengde
print(absorbans[-1])     # siste absorbans
print(absorbans[2:5])    # element 2, 3 og 4
print(bolgelengde[::2])  # annenhver bølgelengde
```

Et utsnitt som `2:5` inkluderer indeks 2, men ikke indeks 5. Dette følger samme mønster som `range`: sluttverdien er ikke med.

### Matematikk med arrayer

Noe av det mest praktiske med arrayer er at vi kan utføre den samme beregningen på alle elementene samtidig. Operasjonene er vanligvis *elementvise*:

```{code-block} Python
import numpy as np

absorbans = np.array([0.12, 0.31, 0.55, 0.42])
blank = np.array([0.02, 0.02, 0.03, 0.02])

korrigert_absorbans = absorbans - blank
dobbel_absorbans = 2 * absorbans
kvadrert_absorbans = absorbans**2

print(korrigert_absorbans)
print(dobbel_absorbans)
print(kvadrert_absorbans)
```

Når to arrayer kombineres, må de vanligvis ha samme form. Et enkelt tall kan derimot kombineres med hele arrayen. Når vi skriver `2 * absorbans`, multipliseres alle elementene med 2.

NumPy har også matematiske funksjoner som virker på hele arrayer:

```{code-block} Python
import numpy as np

H3O = np.array([1E-3, 1E-5, 1E-7, 1E-9, 1E-11])
pH = -np.log10(H3O)

print(pH)
```

### Vektorisering

*Vektorisering* betyr at vi lar en arrayoperasjon utføre den samme beregningen på mange verdier, i stedet for å gå gjennom verdiene én etter én i ei løkke. Vi kunne ha beregnet pH-verdiene slik:

```{code-block} Python
import numpy as np

H3O = [1E-3, 1E-5, 1E-7, 1E-9, 1E-11]
pH = []

for konsentrasjon in H3O:
    pH.append(-np.log10(konsentrasjon))

print(pH)
```

Den vektoriserte versjonen er kortere:

```{code-block} Python
import numpy as np

H3O = np.array([1E-3, 1E-5, 1E-7, 1E-9, 1E-11])
pH = -np.log10(H3O)

print(pH)
```

Vektorisert kode er ofte både raskere og lettere å lese. Løkker er likevel ikke overflødige: De er nyttige når hver verdi må behandles forskjellig, når beregningen avhenger av resultatet fra forrige trinn, eller når vi skal gjenta en mer sammensatt serie med operasjoner.

### Dictionarier i dictionarier

        "atommasse": 1.008,

    },
    "V": {
        "navn": "vanadium",
        "atomnummer": 23,

print(grunnstoffer["V"]["atommasse"])
