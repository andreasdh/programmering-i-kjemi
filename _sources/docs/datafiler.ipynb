## Nye dataoppgaver: 2.11–2.16

Disse filene brukes i {ref}`de nye oppgavene om kjemiske data <nye-dataoppgaver>`. Alle dataene er syntetiske øvingsdata. Last ned filene du trenger, og legg dem i ei mappe som heter `data` ved siden av notebooken din.

|---|---|---|
| 2.11 | {download}`ledningsevne.csv <datahandtering_visualisering/data/ledningsevne.csv>` | NaCl-løsninger: avlesninger, status og dokumenterte datafeil. |
| 2.12 | {download}`pipettekontroll.csv <datahandtering_visualisering/data/pipettekontroll.csv>` | Veiinger ved separate pipetteringer med to pipetter. |
| 2.13 | {download}`klorid_titrering.csv <datahandtering_visualisering/data/klorid_titrering.csv>` | Kloridresultater i mg/L med prøve-ID. |
| 2.13 | {download}`klorid_ionekromatografi.csv <datahandtering_visualisering/data/klorid_ionekromatografi.csv>` | Kloridresultater i mmol/L med prøve-ID. |
| 2.14 | {download}`ph_lagring.csv <datahandtering_visualisering/data/ph_lagring.csv>` | pH-tidsserier for to løsninger. Semikolon og desimalkomma. |
| 2.15 | {download}`natrium_flammeemisjon.csv <datahandtering_visualisering/data/natrium_flammeemisjon.csv>` | Standarder og ukjente prøver med fortynningsfaktorer. |
| 2.16 | {download}`koffein_standarder.csv <datahandtering_visualisering/data/koffein_standarder.csv>` | Koffeinstandarder med kjente konsentrasjoner og topparealer. |
